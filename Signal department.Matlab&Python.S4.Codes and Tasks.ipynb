# **🐍 Welcome to Python for EEG Data!**

**Why Learn Python?**

Python is a popular programming language because it is easy to learn, versatile, and widely used. It has a large collection of libraries for data analysis, visualization, statistics, machine learning, and automation. It is a high-level language such as C++, Java & Pacal but it's syntax is more simple and easier to learn. 

**Why Python for Neuroscience**

Python is particularly useful for neuroscience because it has specialized tools for working with EEG, MEG, fMRI, and other neuroscience data. For example, MNE-Python provides functions for EEG preprocessing, filtering, artifact removal, epoching, time-frequency analysis, and visualization.

Python also makes it easy to combine neuroscience with statistics, signal processing, machine learning, and data visualization.

# How to write code? 

Common traits of a good code include:

**Readability**: Code readability is essential to maintenance, debugging and coordination across teams and time. Can another team member easily understand your code? Can another programmer, working years from now, accurately interpret your code without you being around to provide context?

**Maintainability**: Code complexity is often inversely correlated with code maintainability. Is your code easily testable, enabling your team to efficiently evaluate it for security vulnerabilities and optimization opportunities? Is it robust enough to add new features without breaking core functionality, or is it too brittle to adjust without requiring major refactoring? Prioritizing maintainable code might entail additional upfront hassle, but saves significant time and energy moving forward.

**Efficiency**: Well-written code can reduce a system’s latency and resource consumption. For instance, carefully chosen data structures can minimize the number of CPU operations required for a given function. Thoughtful data caching strategies cut down on costly input/out (I/O) operations by eliminating redundant database queries and network requests, while promptly releasing unused memory avoids unnecessary RAM bloat.

**Reliability**: Fewer defects and structures robust to code changes means less frequent failures and downtime. Reliability is essential to user experience and trust, as well as to the health of critical systems on which your company relies.

#### ✔️Decimal Numbers
There are two types of float based on how much memory they use. 
In Python, the default float is 64-bit (float64).

|              | 32-bit float (`float32`)                      | 64-bit float (`float64`) |
| ------------ | --------------------------------------------- | ------------------------ |
| Memory       | 32 bits = 4 bytes                             | 64 bits = 8 bytes        |
| Precision    | ~7 decimal digits                             | ~15–16 decimal digits    |
| Range        | Smaller                                       | Much larger              |
| Memory usage | Lower                                         | Higher                   |
| Common use   | Large datasets, when precision isn't critical | Scientific/data analysis |


#### Fun fact: 
It's called a float because of the idea of a floating-point number.

The decimal point can "float" to different positions depending on the size of the number.

For example:

x = 12.5

y = 0.125

z = 1250.0

All of these are floating-point numbers. The decimal point can be represented at different positions.

This allows computers to represent both very large and very small numbers.

<h2 style="color:lightcoral;">📌Assignment Operators</h2>
