<h2 style="color:white;"><b>🟧 Containers in Python</b></h2>

### Cognitive Neuroscience — Python
#### The four built-in containers: **List · Dictionary · Tuple · Set**

| Container | Syntax | Ordered | Mutable | Duplicates | Indexable |
|---|---|---|---|---|---|
| **List** | `[ ]` | ✅ | ✅ | ✅ | ✅ |
| **Dictionary** | `{key: value}` | ✅ (3.7+) | ✅ | keys ❌ / values ✅ | by **key** |
| **Tuple** | `( )` | ✅ | ❌ | ✅ | ✅ |
| **Set** | `{ }` | ❌ | ✅ | ❌ | ❌ (use a loop) |

<span style="color:pink; font-size:25px; font-family: 'Calibri', sans-serif;">Gotcha: {} makes an empty DICT, not a set. Use set() for an empty set.</span>

```python
empty_dict = {}
empty_set  = set()
print(type(empty_dict))   # <class 'dict'>
print(type(empty_set))    # <class 'set'>
```

## ➖➖➖➖➖➖➖➖➖➖➖➖➖➖

## **📚 List**

#### **✓ Ordered: keeps the insertion order**
#### **✓ Mutable: items can be changed**
#### **✓ Heterogeneous: can hold mixed data types**
#### **✓ Allows duplicates**

<h2 style="color:lightcoral;">📌 Creating a list</h2>

```python
eeg_channels = ["Fp1", "Fp2", "Cz", "O1", "O2"]   # literal
eeg_info     = ["Fp1", 256, True, 120.5]              # mixed types
from_tuple   = list(("Fp1", "Fp2", "Cz"))            # list() constructor
empty        = list()                                 # empty list
print(eeg_channels)
print(eeg_info)
print(from_tuple, empty)
```

<h2 style="color:lightcoral;">📌 Indexing</h2>

```python
eeg_channels = ["Fp1", "Fp2", "Cz", "O1", "O2"]
print(eeg_channels[0])    # first
print(eeg_channels[-1])   # last (negative index)
print(eeg_channels[-2])   # one before last
```

<h2 style="color:lightcoral;">📌 Slicing:  [:stop] · [start:] · [:] · [::step] · [::-1]</h2>

```python
eeg_channels = ["Fp1", "Fp2", "Cz", "O1", "O2"]
print(eeg_channels[:2])    # first two
print(eeg_channels[2:])    # from index 2 to end
print(eeg_channels[1:4])   # index 1,2,3
print(eeg_channels[:])     # full copy
print(eeg_channels[::2])   # every second item
print(eeg_channels[::-1])  # reversed
```

<h2 style="color:lightcoral;">📌 Membership:  in / not in</h2>

```python
eeg_channels = ["Fp1", "Fp2", "Cz"]
print("Cz" in eeg_channels)       # True
print("Pz" not in eeg_channels)   # True
```

<span style="color:pink; font-size:25px; font-family: 'Calibri', sans-serif;">append(): add ONE item to the end</span>

```python
eeg_channels = ["Fp1", "Fp2"]
eeg_channels.append("Cz")
print(eeg_channels)
```

<span style="color:pink; font-size:25px; font-family: 'Calibri', sans-serif;">extend(): add MANY items from another iterable (append vs extend!)</span>

```python
a = ["Fp1", "Fp2"]
a.extend(["Cz", "O1"])     # each item added separately
print(a)

b = ["Fp1", "Fp2"]
b.append(["Cz", "O1"])     # the WHOLE list added as one item
print(b)
```

<span style="color:pink; font-size:25px; font-family: 'Calibri', sans-serif;">insert(i, x): insert x at position i</span>

```python
eeg_channels = ["Fp1", "Cz"]
eeg_channels.insert(1, "Fp2")
print(eeg_channels)
```

<span style="color:pink; font-size:25px; font-family: 'Calibri', sans-serif;">remove(x): remove first matching value</span>

```python
eeg_channels = ["Fp1", "Fp2", "Cz", "Fp2"]
eeg_channels.remove("Fp2")
print(eeg_channels)
```

<span style="color:pink; font-size:25px; font-family: 'Calibri', sans-serif;">pop(i): remove & RETURN item at i (default last)</span>

```python
eeg_channels = ["Fp1", "Fp2", "Cz"]
last = eeg_channels.pop()
print(last, eeg_channels)
```

<span style="color:pink; font-size:25px; font-family: 'Calibri', sans-serif;">del: delete an item by index, or the whole list</span>

```python
eeg_channels = ["Fp1", "Fp2", "Cz"]
del eeg_channels[0]        # delete one item
print(eeg_channels)
```

<span style="color:pink; font-size:25px; font-family: 'Calibri', sans-serif;">index(x) & count(x): position of first x / how many x</span>

```python
eeg_channels = ["Fp1", "Fp2", "Cz", "Fp2"]
print(eeg_channels.index("Cz"))    # 2
print(eeg_channels.count("Fp2"))   # 2
```

<span style="color:pink; font-size:25px; font-family: 'Calibri', sans-serif;">sort() & sorted(): in-place sort vs. returning a new sorted list</span>

```python
amps = [32.5, 8.2, 40.0, 24.4]
print(sorted(amps))              # returns a NEW sorted list
print(sorted(amps, reverse=True))
print(amps)                      # original unchanged

amps.sort()                      # sorts IN PLACE (changes the list)
print(amps)
```

<span style="color:pink; font-size:25px; font-family: 'Calibri', sans-serif;">reverse(): reverse the list in place</span>

```python
eeg_channels = ["Fp1", "Fp2", "Cz"]
eeg_channels.reverse()
print(eeg_channels)
```

<span style="color:pink; font-size:25px; font-family: 'Calibri', sans-serif;">len(): number of items</span>

```python
eeg_channels = ["Fp1", "Fp2", "Cz", "O1"]
print(len(eeg_channels))
```

<span style="color:pink; font-size:25px; font-family: 'Calibri', sans-serif;">copy(): a shallow copy (independent list)</span>

```python
a = ["Fp1", "Fp2"]
b = a.copy()
b.append("Cz")
print(a, b)
```

<span style="color:pink; font-size:25px; font-family: 'Calibri', sans-serif;">clear(): empty the list</span>

```python
eeg_channels = ["Fp1", "Fp2"]
eeg_channels.clear()
print(eeg_channels)
```

<span style="color:pink; font-size:25px; font-family: 'Calibri', sans-serif;">Concatenation: + joins lists, * repeats</span>

```python
frontal = ["Fp1", "Fp2"]
central = ["Cz", "C3"]
print(frontal + central)
print(frontal * 2)
```

## ➖➖➖➖➖➖➖➖➖➖➖➖➖➖

## **📚 Dictionary**

#### **✓ Stores data as key: value pairs**
#### **✓ Mutable: can add / change / remove pairs**
#### **✓ Keys are unique; values can repeat**
#### **✓ Accessed by KEY (not by position)**

<h2 style="color:lightcoral;">📌 Creating & accessing</h2>

```python
subject = {"id": "S01", "age": 25, "condition": "rest"}   # literal
empty   = dict()                                             # dict() constructor
print(subject["age"])        # access by key
print(subject, empty)
```

<span style="color:pink; font-size:25px; font-family: 'Calibri', sans-serif;">get(): safe access — returns None (or a default) instead of an error</span>

```python
subject = {"id": "S01", "age": 25}
print(subject.get("age"))              # 25
print(subject.get("gender"))           # None (no error)
print(subject.get("gender", "N/A"))    # default value
```

<span style="color:pink; font-size:25px; font-family: 'Calibri', sans-serif;">Add / change a pair:  d[key] = value</span>

```python
amplitudes = {"Fp1": 32, "Fp2": 28}
amplitudes["Cz"] = 40      # add a new pair
amplitudes["Fp1"] = 35     # change an existing value
print(amplitudes)
```

<span style="color:pink; font-size:25px; font-family: 'Calibri', sans-serif;">keys() / values() / items()</span>

```python
amplitudes = {"Fp1": 32, "Fp2": 28, "Cz": 40}
print(amplitudes.keys())
print(amplitudes.values())
print(amplitudes.items())
```

<span style="color:pink; font-size:25px; font-family: 'Calibri', sans-serif;">Membership & looping (a key `in` dict; loop over items)</span>

```python
amplitudes = {"Fp1": 32, "Fp2": 28, "Cz": 40}
print("Cz" in amplitudes)        # checks KEYS -> True
for ch, amp in amplitudes.items():
    print(ch, "->", amp)
```

<span style="color:pink; font-size:25px; font-family: 'Calibri', sans-serif;">pop(key): remove & return that key's value</span>

```python
amplitudes = {"Fp1": 32, "Fp2": 28}
val = amplitudes.pop("Fp1")
print(val, amplitudes)
```

<span style="color:pink; font-size:25px; font-family: 'Calibri', sans-serif;">popitem(): remove & return the LAST inserted pair</span>

```python
amplitudes = {"Fp1": 32, "Fp2": 28}
print(amplitudes.popitem())
print(amplitudes)
```

<span style="color:pink; font-size:25px; font-family: 'Calibri', sans-serif;">del d[key]: delete a specific key</span>

```python
amplitudes = {"Fp1": 32, "Fp2": 28, "Cz": 40}
del amplitudes["Fp2"]
print(amplitudes)
```

<span style="color:pink; font-size:25px; font-family: 'Calibri', sans-serif;">update(): merge another dict / add many pairs</span>

```python
amplitudes = {"Fp1": 32}
amplitudes.update({"Fp2": 28, "Cz": 40})
print(amplitudes)
```

<span style="color:pink; font-size:25px; font-family: 'Calibri', sans-serif;">copy() & clear()</span>

```python
amplitudes = {"Fp1": 32, "Fp2": 28}
backup = amplitudes.copy()
amplitudes.clear()
print(amplitudes, backup)
```

## ➖➖➖➖➖➖➖➖➖➖➖➖➖➖

## **📚 Tuple**

#### **✓ Ordered: keeps insertion order**
#### **✓ Immutable: items CANNOT be changed**
#### **✓ Heterogeneous & allows duplicates**
#### **✓ Faster / safer than a list for fixed data**

<h2 style="color:lightcoral;">📌 Creating a tuple (note the comma for a single item)</h2>

```python
eeg_meta      = ("Fp1", 256, "rest", True)   # mixed types
single        = ("Cz",)                       # ONE item needs a comma
not_a_tuple   = ("Cz")                         # this is just a string!
empty         = tuple()                         # tuple() constructor
from_list     = tuple(["Fp1", "Fp2", "Cz"])   # build from a list
print(type(single), type(not_a_tuple))
print(from_list, empty)
```

<h2 style="color:lightcoral;">📌 Indexing & slicing (same rules as lists)</h2>

```python
eeg_window = (0.0, 0.5, 1.0, 1.5, 2.0)   # seconds
print(eeg_window[0], eeg_window[-1])
print(eeg_window[:2])     # first two
print(eeg_window[3:])     # from index 3
print(eeg_window[::-1])   # reversed
print(eeg_window[1:4:2])  # start:stop:step
```

<h2 style="color:lightcoral;">📌 Membership: in / not in</h2>

```python
eeg_channels = ("Fp1", "Fp2", "Cz", "O1")
print("Cz" in eeg_channels)
print("Pz" not in eeg_channels)
```

<span style="color:pink; font-size:25px; font-family: 'Calibri', sans-serif;">count() / index() / len()</span>

```python
eeg_artifacts = ("blink", "muscle", "blink", "drift")
print(eeg_artifacts.count("blink"))   # 2
print(eeg_artifacts.index("drift"))   # 3
print(len(eeg_artifacts))             # 4
```

<span style="color:pink; font-size:25px; font-family: 'Calibri', sans-serif;">Joining tuples: + concatenates, * repeats</span>

```python
frontal = ("Fp1", "Fp2")
central = ("Cz", "C3", "C4")
print(frontal + central)
print(frontal * 2)
```

<span style="color:pink; font-size:25px; font-family: 'Calibri', sans-serif;">Convert tuple <-> list: the trick to 'edit' an immutable tuple</span>

```python
coords = (0.1, 0.5, 1.2)
coords_list = list(coords)       # tuple -> list
coords_list.append(2.0)          # now editable
back = tuple(coords_list)        # list -> tuple
print(coords_list, back)
```

<span style="color:pink; font-size:25px; font-family: 'Calibri', sans-serif;">Unpacking: assign tuple items to variables at once</span>

```python
eeg_meta = ("Fp1", 256, "rest")
channel, rate, condition = eeg_meta
print(channel, rate, condition)
```

<span style="color:pink; font-size:25px; font-family: 'Calibri', sans-serif;">Immutability: trying to change an item raises TypeError (teaching moment)</span>

```python
eeg_window = (0.0, 0.5, 1.0)
eeg_window[0] = 99   # ❌ TypeError: tuples do not support item assignment
```

## ➖➖➖➖➖➖➖➖➖➖➖➖➖➖

## **📚 Set**

#### **✓ Unordered: no insertion order**
#### **✓ Mutable: items can be added / removed**
#### **✓ Unique: duplicates are dropped automatically**
#### **✓ NOT indexable (use a loop)**

<h2 style="color:lightcoral;">📌 Creating a set (duplicates disappear)</h2>

```python
artifact_types = {"blink", "muscle", "blink", "drift"}   # duplicate 'blink' dropped
from_list      = set(["Fp1", "Fp2", "Fp1"])               # set() -> removes duplicates
print(artifact_types)
print(from_list, len(from_list))
```

<span style="color:pink; font-size:25px; font-family: 'Calibri', sans-serif;">A set is NOT indexable -> loop to access; convert to list for positions</span>

```python
artifact_types = {"blink", "muscle", "drift"}
print(artifact_types[0])   # ❌ TypeError: 'set' object is not subscriptable
```

```python
artifact_types = {"blink", "muscle", "drift"}
for a in artifact_types:      # loop to access elements
    print(a)

as_list = list(artifact_types)   # convert if positional access is needed
print(as_list[0])                # order is NOT guaranteed
```

<h2 style="color:lightcoral;">📌 Membership: in / not in</h2>

```python
artifact_types = {"blink", "muscle", "drift"}
print("blink" in artifact_types)
print("alpha" not in artifact_types)
```

<span style="color:pink; font-size:25px; font-family: 'Calibri', sans-serif;">add(): add one element</span>

```python
artifact_types = {"blink", "muscle"}
artifact_types.add("drift")
artifact_types.add("blink")   # duplicate ignored
print(artifact_types)
```

<span style="color:pink; font-size:25px; font-family: 'Calibri', sans-serif;">update(): add many elements from any iterable</span>

```python
artifact_types = {"blink"}
artifact_types.update(["muscle", "drift"])
print(artifact_types)
```

<span style="color:pink; font-size:25px; font-family: 'Calibri', sans-serif;">remove() vs discard(): remove errors if missing, discard does NOT</span>

```python
s = {"blink", "muscle"}
s.remove("blink")     # removes it
s.discard("alpha")    # not there -> no error (remove would raise KeyError)
print(s)
```

<span style="color:pink; font-size:25px; font-family: 'Calibri', sans-serif;">pop(): remove & return an arbitrary element / clear(): empty it</span>

```python
s = {"blink", "muscle", "drift"}
print(s.pop())   # some element
s.clear()
print(s)
```

<h2 style="color:lightcoral;">📌 Set operations — numeric warm-up (same style as the 30-Days repo)</h2>

```python
A = {19, 22, 24, 20, 25, 26}
B = {19, 28, 26, 27, 20, 25}
print(A.union(B))
print(A.intersection(B))
print(A.difference(B))
print(A.symmetric_difference(B))
```

<span style="color:pink; font-size:25px; font-family: 'Calibri', sans-serif;">EEG example + operators:  | union · & intersection · - difference · ^ sym.diff</span>

```python
method_A = {"blink", "muscle", "drift"}
method_B = {"blink", "drift", "line_noise", "sweat"}
print(method_A | method_B)   # union         (اجتماع)
print(method_A & method_B)   # intersection  (اشتراک)
print(method_A - method_B)   # difference    (تفاضل)
print(method_A ^ method_B)   # symmetric difference
```

<span style="color:pink; font-size:25px; font-family: 'Calibri', sans-serif;">Relationships: issubset / issuperset / isdisjoint</span>

```python
artifacts = {"blink", "muscle", "drift"}
print({"blink"}.issubset(artifacts))
print(artifacts.issuperset({"blink"}))
print(artifacts.isdisjoint({"alpha"}))
```
