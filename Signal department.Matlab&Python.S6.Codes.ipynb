## ➖➖➖➖➖➖➖➖➖➖➖➖➖➖➖

## **🔆 Comparison &amp; Logical Operators**
#### *Before we can make decisions, we need to ask True/False questions.*

````python
a = 8

print(a == 8)   # equal to?
print(a != 5)   # not equal to?
print(a > 10)   # greater than?
print(a < 10)   # less than?
print(a >= 8)   # greater than or equal?
print(a <= 7)   # less than or equal?
````

### **➡️ Combining questions with `and`, `or`, `not`**

````python
freq = 10  # Hz  (an EEG frequency)

print(freq > 8 and freq < 13)   # BOTH must be True  -> in the Alpha band
print(freq < 4 or freq > 30)    # at least ONE must be True
print(not (freq == 10))         # reverses True/False
````

### **➡️ More question types: `in`, chained comparisons**

````python
channels = ["Fp1", "C3", "O2"]

print("C3" in channels)        # membership: is it in the list?
print("Pz" not in channels)    # not in the list?
````

## ➖➖➖➖➖➖➖➖➖➖➖➖➖➖➖

## **🔆 if Statement**
#### *An `if` lets your code choose a path depending on a condition.*

````python
amplitude = 120  # microvolts (µV)

if amplitude > 100:
    print("Possible artifact detected")
````

### **➡️ Adding an `else`**
#### *`else` runs when the `if` condition is False.*

````python
amplitude = 80

if amplitude > 100:
    print("Possible artifact detected")
else:
    print("Signal looks normal")
````

### **➡️ Multiple Conditions**
<h3 style="color:lightblue;"><b>🔹 if – the first condition</b></h3>
<h3 style="color:lightblue;"><b>🔹 elif – any additional condition</b></h3>
<h3 style="color:lightblue;"><b>🔹 else – the default, when nothing above is True</b></h3>

````python
freq = 10  # Hz  -> the dominant EEG frequency

if freq < 4:
    print("Delta band  (deep sleep)")
elif freq < 8:
    print("Theta band  (drowsy / meditation)")
elif freq < 13:
    print("Alpha band  (relaxed, eyes closed)")
elif freq < 30:
    print("Beta band   (alert, active thinking)")
else:
    print("Gamma band  (high-level processing)")
````

### **➡️ Nested `if` — a decision inside a decision**

````python
channel = "Fp1"
amplitude = 130

if channel.startswith("Fp"):              # is it a frontal channel?
    print("Frontal channel")
    if amplitude > 100:
        print("  ...but amplitude is high -> likely an eye-blink artifact")
    else:
        print("  ...amplitude looks fine")
````

### **➡️ More decision examples**

````python
impedance = 8  # kΩ  (electrode contact quality)

if impedance < 5:
    print("Excellent electrode contact")
elif impedance < 10:
    print("Acceptable contact")
else:
    print("Poor contact - re-gel the electrode")
````

````python
rt = 420  # reaction time in ms

if rt < 200:
    print("Too fast - likely anticipatory (invalid)")
elif rt <= 800:
    print("Valid response")
else:
    print("Too slow - possible lapse of attention")
````

### **⚠️ Common Mistake: `=` vs `==`**
#### *`=` **assigns** a value. `==` **compares** two values.*
#### *Run the next cell to SEE the error, then look at the fixed version.*

````python
amplitude = 100

# The line below tries to COMPARE with a single '=' -> SyntaxError
if amplitude = 100:
    print("equal")
````

````python
amplitude = 100

if amplitude == 100:   # correct: '==' compares
    print("amplitude is exactly 100")
````

## **🔆 for Loop**
#### *A `for` loop repeats code once for each item in a sequence (list, string, range, ...).*

````python
# Doing it by hand is tedious and error-prone:
print("Processing EEG sample")
print("Processing EEG sample")
print("Processing EEG sample")
````

````python
# A loop does the same thing, cleanly:
for i in range(3):
    print("Processing EEG sample", i)

# range(n) -> 0, 1, 2, ..., n-1
````

````python
# range(start, stop, step)
for freq in range(0, 40, 5):
    print(f"Checking {freq} Hz")
````

### **➡️ Looping over different sequences**

````python
channels = ["Fp1", "Fp2", "C3", "C4", "O1", "O2"]
for ch in channels:
    print(f"Analyzing channel {ch}")
````

````python
for letter in "EEG":
    print(letter)
````

````python
metadata = {"subject": "S01", "sampling_rate": 256, "channels": 6}

for key in metadata:
    print(key, "->", metadata[key])
````

````python
# .items() gives you the key AND the value together
for key, value in metadata.items():
    print(f"{key}: {value}")
````

### **➡️ Handy loop helpers: `enumerate` and `zip`**

````python
channels = ["Fp1", "C3", "O2"]

# enumerate -> gives you a counter alongside each item
for index, ch in enumerate(channels):
    print(f"Channel #{index} is {ch}")
````

````python
channels    = ["Fp1", "C3", "O2"]
amplitudes  = [90, 45, 60]

# zip -> walk through two lists in parallel
for ch, amp in zip(channels, amplitudes):
    print(f"{ch}: {amp} µV")
````

### **➡️ The accumulator pattern (sum, count, average)**
#### *This pattern shows up constantly in data analysis.*

````python
amplitudes = [12.5, 85.3, 60.1, 95.6, 45.3]

total = 0
for amp in amplitudes:
    total = total + amp          # keep adding into 'total'

average = total / len(amplitudes)
print("Total:  ", total)
print("Average:", average)
````

````python
amplitudes = [12.5, 85.3, 140.2, 95.6, 200.4, 45.3]

artifact_count = 0
for amp in amplitudes:
    if amp > 100:
        artifact_count += 1      # shorthand for artifact_count = artifact_count + 1

print("Number of possible artifacts:", artifact_count)
````

### **➡️ Building a NEW list while looping (the append pattern)**

````python
amplitudes = [12.5, 85.3, 140.2, 95.6, 200.4]

artifacts = []                  # start with an empty list
for amp in amplitudes:
    if amp > 100:
        artifacts.append(amp)   # grow the list as we go

print("Artifacts found:", artifacts)
````

````python
channel_amps = {"Fp1": 90, "C3": 45, "O2": 60, "Pz": 130}

for ch, amp in channel_amps.items():
    if amp > 100:
        print(f"{ch}: {amp} µV  <- check this channel")
    else:
        print(f"{ch}: {amp} µV  ok")
````

### **➡️ Nested loops**

````python
channels = ["Fp1", "C3"]
windows  = [1, 2, 3]

for ch in channels:
    for w in windows:
        print(f"Channel {ch}, time window {w}")
````

````python
amps = [3, 1, 2]
for a in sorted(amps):          # loop in sorted order
    print(a)

print("---")

for ch in reversed(["Fp1", "C3", "O2"]):   # loop in reverse
    print(ch)
````

## **🔆 while Loop**
#### *A `while` loop keeps running as long as its condition stays True.*

````python
x = 0
while x < 5:
    print("sample", x)
    x += 1          # ALWAYS move toward the stop condition
````

### **⚠️ Infinite loops**
#### *If the condition never becomes False, the loop runs forever.*

```python
# DON'T RUN THIS — x never changes, so the loop never stops:
x = 0
while x < 5:
    print("stuck!")
    # we forgot  x += 1
```
#### *If you ever get stuck in a running cell, press the ⏹️ Stop / Interrupt button.*

````python
sample_index = 0
while sample_index < 5:
    print(f"Streaming EEG sample {sample_index}")
    sample_index += 1
````

### **➡️ More `while` patterns**

````python
countdown = 3
while countdown > 0:
    print("beep", countdown)
    countdown -= 1
print("go!")
````

````python
# Keep collecting samples until the running total passes a target
total = 0
sample = 0
while total < 300:
    sample += 10
    total += sample
    print(f"added {sample}, running total = {total}")
print("Target reached.")
````

### **🔄 break &amp; continue**
<h3 style="color:lightblue;"><b>🔹 continue – skip the rest of THIS iteration, jump to the next one</b></h3>
<h3 style="color:lightblue;"><b>🔹 break – stop the loop completely</b></h3>

````python
amplitudes = [12.5, -5.0, 85.3, 0, 60.1]

for amp in amplitudes:
    if amp <= 0:
        print("noise -> skipped")
        continue                 # go straight to the next value
    print("valid amplitude:", amp)
````

````python
amplitudes = [12.5, 85.3, 140.2, 60.1]

for amp in amplitudes:
    if amp > 100:
        print("Artifact detected! Stopping.")
        break                    # leave the loop right now
    print("processed:", amp)
````

### **➡️ Two more everyday uses of `break` and `continue`**

````python
amplitudes = [20, 45, 60, 130, 80]

# find the FIRST artifact, then stop searching
for i, amp in enumerate(amplitudes):
    if amp > 100:
        print(f"First artifact at index {i} (value {amp})")
        break
````

````python
channels = ["Fp1", "EOG", "C3", "EOG", "O2"]

for ch in channels:
    if ch == "EOG":        # skip the eye-movement channel
        continue
    print("Analyzing", ch)
````

````python
# A classic pattern: loop forever, and break when a condition is met.
threshold = 100

while True:
    amp = float(input("Enter EEG amplitude: "))
    if amp > threshold:
        print("Artifact detected!")
        break
    else:
        print("Normal EEG signal")
````

## **🔆 try / except — Handling Errors**
#### *What if the user types text where we expected a number? Python raises an error and the program crashes.*
#### *`try / except` lets us catch the error and respond politely instead.*

### **⚠️ First, let's SEE the error**

````python
# float() cannot turn "abc" into a number -> ValueError
amp = float("abc")
````

````python
# Now we catch it:
try:
    amp = float("abc")
    print("amplitude:", amp)
except ValueError:
    print("Invalid input! Please enter a numeric value.")
````

### **➡️ Other common errors you can catch**

````python
amplitudes = []

try:
    average = sum(amplitudes) / len(amplitudes)   # dividing by zero (empty list)!
    print(average)
except ZeroDivisionError:
    print("No samples to average.")
````

````python
metadata = {"subject": "S01", "sampling_rate": 256}

try:
    print(metadata["age"])          # this key does not exist
except KeyError:
    print("That field is missing from the metadata.")
````

#### *`else` runs if there was no error; `finally` runs no matter what:*

````python
value = "50"

try:
    amp = float(value)
except ValueError:
    print("Not a number")
else:
    print("Converted successfully:", amp)   # only if no error
finally:
    print("Done checking.")                 # always runs
````

### 📌 Function with Parameters

````python
def inspect_channel(channel):
    print(f"Inspecting EEG channel {channel}...")

inspect_channel("Fp1")
inspect_channel("C3")
````

````python
def label_recording(subject_id, session_id):
    print(subject_id + " - " + session_id)

label_recording("S01", "Session1")
````

#### *More parameter examples:*

````python
def mv_to_uv(mv):
    return mv * 1000        # 1 millivolt = 1000 microvolts

print(mv_to_uv(0.05), "µV")
print(mv_to_uv(0.12), "µV")
````

````python
def describe_channel(name, amplitude):
    return f"Channel {name} has amplitude {amplitude} µV"

print(describe_channel("Fp1", 90))
print(describe_channel("C3", 45))
````

### 📌 return — sending a value BACK
#### 🟡 *`print()` just shows text on the screen.*
#### 🟡 *`return` hands a value back so you can store it and keep using it.*

````python
def add(a, b):
    return a + b

result = add(6, 7)
print("The sum is:", result)

# because add() RETURNS a value, we can keep computing with it:
print(add(6, 7) + 10)
````

### **⚠️ print vs return — the classic trap**
#### *A function that only `print`s gives back `None`.*

````python
def add_print(a, b):
    print(a + b)              # only prints; returns nothing (None)

value = add_print(6, 7)       # prints 13
print("stored value:", value) # -> None !
print(value + 10)             # -> TypeError: can't add int to None
````

````python
def add_return(a, b):
    return a + b              # returns the value

value = add_return(6, 7)
print("stored value:", value) # 13
print(value + 10)             # 23
````

### 📌 The payoff: rewrite our band-classifier as ONE reusable function

````python
def classify_band(freq):
    if freq < 4:
        return "Delta"
    elif freq < 8:
        return "Theta"
    elif freq < 13:
        return "Alpha"
    elif freq < 30:
        return "Beta"
    else:
        return "Gamma"

print(classify_band(10))   # Alpha
print(classify_band(2))    # Delta

# now reuse it inside a loop — write once, use many times:
for f in [2, 6, 10, 22, 40]:
    print(f, "Hz ->", classify_band(f))
````

#### *A few more functions that `return` useful values:*

````python
def mean_amplitude(amplitudes):
    return sum(amplitudes) / len(amplitudes)

print(mean_amplitude([12.5, 85.3, 60.1, 95.6]))
````

````python
def is_artifact(amplitude, threshold=100):
    return amplitude > threshold      # returns True or False

print(is_artifact(120))
print(is_artifact(80))
````

### 📌 Default Parameters

````python
def analyze_channel(channel="Fp1"):
    print(f"Analyzing EEG channel {channel}...")

analyze_channel()        # uses the default ("Fp1")
analyze_channel("C3")    # overrides the default
````

````python
def flag_artifact(amplitude, threshold=100):
    if amplitude > threshold:
        return "Artifact"
    return "Normal"

print(flag_artifact(120))         # uses threshold = 100
print(flag_artifact(120, 150))    # custom threshold = 150
````

````python
def scale_signal(amplitude, factor=2):
    return amplitude * factor

print(scale_signal(50))       # factor defaults to 2
print(scale_signal(50, 3))    # factor = 3
````

### **Is it possible to create a function without giving it a name?** 🤔

<h2 style="color:lightcoral;">📌 lambda</h2>

#### *A lambda is a small, anonymous function.*
#### *It can take any number of arguments, but holds only ONE expression.*

### **lambda arguments : expression**

````python
# def version
def square_def(x):
    return x * x

# lambda version (anonymous, one expression)
square = lambda x: x * x

print(square_def(5))
print(square(5))
````

````python
add  = lambda a, b: a + b
diff = lambda a, b, c: a + b - c

print(add(2, 3))
print(diff(5, 6, 2))
````

#### *A lambda can hold a small `if/else` too:*

````python
to_status = lambda amp: "Artifact" if amp > 100 else "Normal"

print(to_status(120))
print(to_status(80))
````

### **Lambdas are often used with:**
#### ✔️ `map()` — apply a function to every item
#### ✔️ `filter()` — keep items that match a condition
#### ✔️ `sorted()` — custom sort order

<h3 style="color:white;"> ☑️ map()</h3>

#### map(function, iterable)

````python
nums = [1, 2, 3, 4]
squared = list(map(lambda x: x**2, nums))
print(squared)
````

````python
channels = ['fp1', 'c3', 'o2']
upper = list(map(lambda ch: ch.upper(), channels))
print(upper)
````

````python
mv = [0.02, 0.05, 0.1]
uv = list(map(lambda x: x * 1000, mv))   # convert every mV value to µV
print(uv)
````

````python
a = [10, 20, 30]
b = [1, 2, 3]
sums = list(map(lambda x, y: x + y, a, b))   # map can walk TWO lists at once
print(sums)
````

<h3 style="color:white;"> ☑️ filter()</h3>

#### filter(function, iterable)

````python
nums = [1, 2, 3, 4, 5, 6]
evens = list(filter(lambda x: x % 2 == 0, nums))
print(evens)
````

````python
amplitudes = [-5.0, 12.5, 0, 85.3, -2.1, 60.1]
valid = list(filter(lambda a: a > 0, amplitudes))   # keep only positive amplitudes
print(valid)
````

````python
amplitudes = [12.5, 130.0, 85.3, 200.4, 60.1]
artifacts = list(filter(lambda a: a > 100, amplitudes))   # keep only artifacts
print(artifacts)
````

````python
channels = ["Fp1", "EOG", "C3", "EMG", "O2"]
eeg_only = list(filter(lambda c: c not in ("EOG", "EMG"), channels))
print(eeg_only)
````

<h3 style="color:white;"> ☑️ sorted()</h3>

#### sorted(iterable, key=function, reverse=False)
#### *`key` tells sorted() WHAT to sort by.*

````python
channels = [
    {'channel': 'Fp1', 'amplitude': 90},
    {'channel': 'C3',  'amplitude': 85},
    {'channel': 'O2',  'amplitude': 95},
]

by_amp = sorted(channels, key=lambda x: x['amplitude'])
print(by_amp)
````

````python
channels = ['Fp1', 'T3', 'Pz', 'O2', 'Fpz']
print(sorted(channels, key=lambda ch: len(ch)))     # sort by name length

pairs = [(1, 3), (2, 1), (4, 2)]
print(sorted(pairs, key=lambda p: p[1]))            # sort by the 2nd item
````

````python
channels = [
    {'channel': 'Fp1', 'amplitude': 90},
    {'channel': 'C3',  'amplitude': 85},
    {'channel': 'O2',  'amplitude': 95},
]

top = sorted(channels, key=lambda x: x['amplitude'], reverse=True)
print("highest first:", top)
````

#### **Task 2:**

✔️ Write a function that takes two EEG signal values and an operation symbol (e.g., '+').

✔️ Use a **lambda** to perform the operation and return the result (handle at least 3 operations):
- '+' → combine two amplitudes
- '-' → amplitude difference
- '*' → scale signal intensity

````python

def eeg_operation(signal1, signal2, operation):

    if operation == "+":
        calculate = lambda x, y: x + y
    elif operation == "-":
        calculate = lambda x, y: x - y
    elif operation == "*":
        calculate = lambda x, y: x * y
    else:
        return "Invalid operation"

    return calculate(signal1, signal2)


# Test the function
print(eeg_operation(20, 10, "+"))
print(eeg_operation(20, 10, "-"))
print(eeg_operation(20, 10, "*"))
print(eeg_operation(20, 10, "/"))
````

#### **Task 3:**

✔️ Write a function that takes a list of tuples `(channel_name, average_amplitude)` and:
- sorts the channels by amplitude
- keeps only channels with amplitude &gt; 70 µV
- returns the final list

````python

def filter_eeg_channels(channels):
    sorted_channels = sorted(channels, key=lambda channel: channel[1])

    filtered_channels = [
        channel for channel in sorted_channels
        if channel[1] > 70
    ]

    return filtered_channels


# Test the function
eeg_channels = [
    ("Fz", 65),
    ("Cz", 85),
    ("Pz", 72),
    ("Oz", 50),
    ("F3", 95)
]

result = filter_eeg_channels(eeg_channels)
print(result)
````

#### **Task 4:**

✔️ Write a function that takes a list of EEG amplitudes and:
- removes all noise values (amplitude ≤ 0)
- computes the average of the remaining amplitudes
- returns "High overall activity" if the average &gt; 50 µV, otherwise "Normal overall activity"

**Hint:** handle the case where every value is noise.

````python

def analyze_eeg_activity(amplitudes):
    valid_amplitudes = [
        amplitude for amplitude in amplitudes
        if amplitude > 0
    ]

    if len(valid_amplitudes) == 0:
        return "No valid EEG amplitudes"

    average = sum(valid_amplitudes) / len(valid_amplitudes)

    if average > 50:
        return "High overall activity"
    else:
        return "Normal overall activity"


# Test the function
print(analyze_eeg_activity([20, 40, 60, 80, 0, -10]))
print(analyze_eeg_activity([10, 20, 30]))
print(analyze_eeg_activity([0, -5, -10]))
````

## **🏆 Capstone — Putting It All Together**

Combine **control flow + functions** in one program.

✔️ Write a function `analyze_session(amplitudes)` that receives a list of EEG amplitudes and:
1. skips noise (amplitude ≤ 0) using a loop + `continue`
2. counts how many values are possible artifacts (&gt; 100 µV)
3. computes the average of the valid, non-artifact amplitudes
4. returns a small summary dictionary, e.g.
   `{"valid": ..., "artifacts": ..., "average": ...}`

Then loop over several subjects' data and print a report for each one.

```python
subjects = {
    "S01": [12.5, -5.0, 85.3, 140.2, 60.1, 0, 95.6],
    "S02": [30.0, 45.5, -1.0, 110.7, 20.2],
}
```

````python

def analyze_session(amplitudes):
    valid_amplitudes = []
    artifacts = 0

    for amplitude in amplitudes:
        if amplitude <= 0:
            continue

        if amplitude > 100:
            artifacts += 1
            continue

        valid_amplitudes.append(amplitude)

    if len(valid_amplitudes) > 0:
        average = sum(valid_amplitudes) / len(valid_amplitudes)
    else:
        average = None

    return {
        "valid": len(valid_amplitudes),
        "artifacts": artifacts,
        "average": average
    }


subjects = {
    "S01": [12.5, -5.0, 85.3, 140.2, 60.1, 0, 95.6],
    "S02": [30.0, 45.5, -1.0, 110.7, 20.2],
}

for subject, amplitudes in subjects.items():
    result = analyze_session(amplitudes)

    print(f"Subject: {subject}")
    print(f"Valid amplitudes: {result['valid']}")
    print(f"Artifacts: {result['artifacts']}")
    print(f"Average amplitude: {result['average']}")
    print("-" * 30)
````
