#### **Task 1 — EEG Band Classification**

<span style="color:pink; font-size:25px; font-family: 'Calibri', sans-serif;">نسخه‌ی اصلی (با input)</span>

```python
freq = float(input("Enter a frequency (Hz): "))

if freq < 0.5:
    print("Below EEG range")
elif freq <= 4:
    print("Delta")
elif freq <= 8:
    print("Theta")
elif freq <= 13:
    print("Alpha")
elif freq <= 30:
    print("Beta")
else:
    print("Gamma")
```

<span style="color:pink; font-size:25px; font-family: 'Calibri', sans-serif;">نسخه‌ی قابل‌اجرا (تابع + حلقه‌ی نمونه)</span>

```python
def classify_band(freq):
    if freq < 0.5:   return "Below EEG range"
    elif freq <= 4:  return "Delta"
    elif freq <= 8:  return "Theta"
    elif freq <= 13: return "Alpha"
    elif freq <= 30: return "Beta"
    else:            return "Gamma"

for f in [2, 6, 10, 22, 40]:
    print(f, "Hz ->", classify_band(f))
```

```python
threshold = 100
while True:
    amp = float(input("Enter EEG amplitude (µV): "))
    if amp > threshold:
        print("Artifact detected!")
        break
    else:
        print("Normal signal, keep going...")
```

#### **Task 4 — Identify Noisy EEG Segments**

```python
eeg_amplitudes = [0.5, 3.2, -1.0, 12.4, 18.5, 0, 7.8, 21.3, 5.6, -0.7]

for amp in eeg_amplitudes:
    if amp <= 0:
        print("noise detected, skipped")
        continue
    elif amp < 10:
        print("low EEG activity")
    elif amp <= 20:
        print("high EEG activity")
    else:
        print("extreme amplitude! stopping recording")
        break
```

#### **Task 1 — Amplitude range checker**

```python
def check_amplitude(amp):
    if amp <= 0:
        return "noise or flat signal"
    elif amp <= 100:
        return "valid"
    else:
        return "artifact"

print(check_amplitude(50))    # valid
print(check_amplitude(-3))    # noise or flat signal
print(check_amplitude(150))   # artifact
```

#### **Task 2 — Lambda calculator**

```python
def signal_op(a, b, op):
    ops = {
        "+": lambda x, y: x + y,   # combine amplitudes
        "-": lambda x, y: x - y,   # amplitude difference
        "*": lambda x, y: x * y,   # scale intensity
    }
    if op not in ops:
        return "unsupported operation"
    return ops[op](a, b)

print(signal_op(60, 40, "+"))   # 100
print(signal_op(60, 40, "-"))   # 20
print(signal_op(60, 2, "*"))    # 120
```

#### **Task 3 — Sort + filter channels**

```python
def top_channels(pairs):
    strong = [p for p in pairs if p[1] > 70]     # keep amplitude > 70
    return sorted(strong, key=lambda p: p[1])    # sort by amplitude

data = [("Fp1", 90), ("C3", 45), ("O2", 72), ("Pz", 130), ("T3", 60)]
print(top_channels(data))
# [('O2', 72), ('Fp1', 90), ('Pz', 130)]
```

#### **Task 4 — Clean + average + verdict**

```python
def overall_activity(amplitudes):
    clean = [a for a in amplitudes if a > 0]     # drop noise
    if not clean:                                # guard: all noise
        return "No valid signal (all noise)"
    avg = sum(clean) / len(clean)
    return "High overall activity" if avg > 50 else "Normal overall activity"

print(overall_activity([60, 70, 80]))      # High overall activity
print(overall_activity([10, 20, -5, 0]))   # Normal overall activity
print(overall_activity([-1, 0, -3]))       # No valid signal (all noise)
```

## **🏆 Capstone — Putting It All Together**

```python
def analyze_session(amplitudes):
    valid = 0
    artifacts = 0
    keep = []                       # valid, non-artifact amplitudes
    for amp in amplitudes:
        if amp <= 0:
            continue                # skip noise
        valid += 1
        if amp > 100:
            artifacts += 1          # possible artifact
        else:
            keep.append(amp)
    average = sum(keep) / len(keep) if keep else 0
    return {"valid": valid, "artifacts": artifacts, "average": round(average, 2)}

subjects = {
    "S01": [12.5, -5.0, 85.3, 140.2, 60.1, 0, 95.6],
    "S02": [30.0, 45.5, -1.0, 110.7, 20.2],
}

for sid, amps in subjects.items():
    print(sid, "->", analyze_session(amps))
# S01 -> {'valid': 5, 'artifacts': 1, 'average': 63.38}
# S02 -> {'valid': 4, 'artifacts': 1, 'average': 31.9}
```
